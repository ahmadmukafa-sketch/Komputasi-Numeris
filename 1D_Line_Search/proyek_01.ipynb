```python
import math

#FUNGSI DAN TURUNANNYA
def f(x):
    """Fungsi objektif: f(x) = x^2 - sin(x)"""
    return x**2 - math.sin(x)

def df(x):
    """Turunan pertama: f'(x) = 2x - cos(x)"""
    return 2*x - math.cos(x)

def ddf(x):
    """Turunan kedua: f''(x) = 2 + sin(x)"""
    return 2 + math.sin(x)
```

```python
a_init, b_init = 0.0, 2.0
target_interval = 0.2
```

```python
#GOLDEN SECTION METHOD
a, b = a_init, b_init
rho = (3 - math.sqrt(5)) / 2  #Konstanta Golden Ratio
iter_golden = 0

while (b - a) > target_interval:
    iter_golden += 1
    a1 = a + rho * (b - a)
    b1 = a + (1 - rho) * (b - a)
    
    if f(a1) < f(b1):
        b = b1
    else:
        a = a1

print(f"1. Golden Section Method")
print(f"   Interval Akhir: [{a:.4f}, {b:.4f}] (Lebar: {b-a:.4f})")
print(f"   Jumlah Iterasi: {iter_golden}\n")
```

```python
#FIBONACCI METHOD
a, b = a_init, b_init
F = [1, 1, 2, 3, 5, 8, 13, 21, 34] #Deret Fibonacci
N = 5 
iter_fibo = 0

for k in range(1, N + 1):
    iter_fibo += 1
    rho_k = 1 - (F[N - k + 1] / F[N - k + 2]) if k < N else 0.5 - 0.01
    
    a1 = a + rho_k * (b - a)
    b1 = a + (1 - rho_k) * (b - a)
    
    if f(a1) < f(b1):
        b = b1
    else:
        a = a1

print(f"2. Fibonacci Method")
print(f"   Interval Akhir: [{a:.4f}, {b:.4f}] (Lebar: {b-a:.4f})")
print(f"   Jumlah Iterasi: {iter_fibo}\n")
```

```python
#BISECTION METHOD
a, b = a_init, b_init
iter_bisect = 0

while (b - a) > target_interval:
    iter_bisect += 1
    x0 = (a + b) / 2
    
    # Menggunakan turunan pertama (f')
    if df(x0) > 0:
        b = x0
    else:
        a = x0

print(f"3. Bisection Method")
print(f"   Interval Akhir: [{a:.4f}, {b:.4f}] (Lebar: {b-a:.4f})")
print(f"   Jumlah Iterasi: {iter_bisect}\n")
```

```python
#NEWTON'S METHOD
x = (a_init + b_init) / 2  # Titik awal di tengah interval
iter_newton = 0

while True:
    iter_newton += 1
    x_baru = x - (df(x) / ddf(x))
    
    if abs(x_baru - x) <= target_interval:
        x = x_baru
        break
    x = x_baru

print(f"4. Newton's Method")
print(f"   Estimasi Akar x*: {x:.4f}")
print(f"   Jumlah Iterasi: {iter_newton}\n")
```

```python
#SECANT METHOD
x_prev = 0.0
x_curr = 2.0
iter_secant = 0

while True:
    iter_secant += 1
    # Menggunakan turunan pertama
    df_curr = df(x_curr)
    df_prev = df(x_prev)
    
    x_baru = x_curr - df_curr * (x_curr - x_prev) / (df_curr - df_prev)
    
    if abs(x_baru - x_curr) <= target_interval:
        x_curr = x_baru
        break
        
    x_prev = x_curr
    x_curr = x_baru

print(f"5. Secant Method")
print(f"   Estimasi Akar x*: {x_curr:.4f}")
print(f"   Jumlah Iterasi: {iter_secant}\n")
```

```python
print("=== KESIMPULAN JUMLAH ITERASI ===")
print(f"Golden Section : {iter_golden} iterasi")
print(f"Fibonacci      : {iter_fibo} iterasi")
print(f"Bisection      : {iter_bisect} iterasi")
print(f"Newton's       : {iter_newton} iterasi")
print(f"Secant         : {iter_secant} iterasi")
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([0.24, 0.65, 0.95, 1.24, 1.73, 2.01, 2.23, 2.52])
y = np.array([0.23, -0.23, -1.1, -0.45, 0.27, 0.1, -0.29, 0.24])
n = len(x)
```

```python
X_linear = np.column_stack((np.ones(n), x))
A_linear = np.linalg.inv(X_linear.T @ X_linear) @ X_linear.T @ y 
y_pred_linear = X_linear @ A_linear
sse_linear = np.sum((y - y_pred_linear)**2)
print(f"--- REGRESI LINEAR ---")
print(f"Persamaan: y = {A_linear[0]:.4f} + {A_linear[1]:.4f}x")
print(f"Total Error (SSE): {sse_linear:.4f}\n")
```

```python
X_quad = np.column_stack((np.ones(n), x, x**2))
A_quad = np.linalg.inv(X_quad.T @ X_quad) @ X_quad.T @ y
y_pred_quad = X_quad @ A_quad
sse_quad = np.sum((y - y_pred_quad)**2)
print(f"--- REGRESI QUADRATIC ---")
print(f"Persamaan: y = {A_quad[0]:.4f} + {A_quad[1]:.4f}x + {A_quad[2]:.4f}x^2")
print(f"Total Error (SSE): {sse_quad:.4f}\n")
```

```python
x_line = np.linspace(min(x) - 0.2, max(x) + 0.2, 100)
y_line_linear = A_linear[0] + A_linear[1]*x_line
y_line_quad = A_quad[0] + A_quad[1]*x_line + A_quad[2]*(x_line**2)
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='black', label='Data (Slide 31)', zorder=5)
plt.plot(x_line, y_line_linear, color='red', linestyle='--', label='Linear Regression')
plt.plot(x_line, y_line_quad, color='blue', label='Quadratic Regression')
plt.title('Least Squares Curve Fitting: Linear vs Quadratic')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
print("=== KESIMPULAN ===")
if sse_quad < sse_linear:
    print("Regresi QUADRATIC lebih baik karena menghasilkan total error (SSE) yang lebih kecil.")
else:
    print("Regresi LINEAR lebih baik karena menghasilkan total error (SSE) yang lebih kecil.")
```
